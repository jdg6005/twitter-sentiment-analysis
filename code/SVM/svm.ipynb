```python
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, confusion_matrix,precision_score, recall_score, f1_score
from sklearn.model_selection import train_test_split
import sklearn.metrics as metrics
```

```python
df = pd.read_csv("../../data/twitter_training.csv", header=None) # path to data
df.columns = ["id", "entity", "sentiment", "tweet"] # naming columns for easier access
df = df.dropna()
print(df["tweet"])

x = df.drop(columns="sentiment")
y = df["sentiment"]

x_train, x_val, y_train, y_val = train_test_split(x, y, test_size=0.2, random_state=35, stratify=y)
```

```python
 # min_df removes words appearing in less than n documents
 # max_df removes words appearing in more than n% of documents
 # stop_words removes common english stop words
vectorizer = TfidfVectorizer(ngram_range=(2, 2), min_df=2, max_df=0.8, stop_words='english')
train_tfidf_matrix = vectorizer.fit_transform(x_train['tweet'])
val_tfidf_matrix = vectorizer.transform(x_val["tweet"])
```

```python
svm1 = SVC(kernel="linear", C = 0.1)
svm2 = SVC(kernel="linear", C = 0.5)
svm3 = SVC(kernel="linear", C = 1.0)
svm1.fit(train_tfidf_matrix, y_train)
svm2.fit(train_tfidf_matrix, y_train)
svm3.fit(train_tfidf_matrix, y_train)
```

```python
preds = svm1.predict(val_tfidf_matrix)
acc = metrics.accuracy_score(y_val, preds)
precision = metrics.precision_score(y_val, preds, average='weighted')
recall = metrics.recall_score(y_val, preds, average='weighted')
f1 = metrics.f1_score(y_val, preds, average='weighted')
print("Accuracy", acc)
print("Precision", precision)
print("Recall", recall)
print("f1", f1)
print(preds)
```

```python
preds = svm2.predict(val_tfidf_matrix)
acc = metrics.accuracy_score(y_val, preds)
precision = metrics.precision_score(y_val, preds, average='weighted')
recall = metrics.recall_score(y_val, preds, average='weighted')
f1 = metrics.f1_score(y_val, preds, average='weighted')
print("Accuracy", acc)
print("Precision", precision)
print("Recall", recall)
print("f1", f1)
```

```python
preds = svm3.predict(val_tfidf_matrix)
acc = metrics.accuracy_score(y_val, preds)
precision = metrics.precision_score(y_val, preds, average='weighted')
recall = metrics.recall_score(y_val, preds, average='weighted')
f1 = metrics.f1_score(y_val, preds, average='weighted')
print("Accuracy", acc)
print("Precision", precision)
print("Recall", recall)
print("f1", f1)
```
